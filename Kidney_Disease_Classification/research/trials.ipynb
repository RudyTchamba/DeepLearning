**Ensure Annotation**

Here it is used to ensure that when defining the types of integers that are to be accepted by a function t is exactly those types that the function only accepts

```python
def get_predict(x:int, y:int) -> int:
    return x * y
```

```python
get_predict(2, 3)
```

```python
get_predict(2, "3")
```

**using the ensure anotations**

```python
from ensure import ensure_annotations
```

```python
@ensure_annotations
def get_predict(x:int, y:int) -> int:
    return x * y
```

```python
get_predict(2, 3)
```

```python
get_predict(2, "3")
```
